```python
#You can guide the behavior of Gemini models with system instructions. To do so, pass a GenerateContentConfig object.

from google import genai
from google.genai import types

client = genai.Client()

response = client.models.generate_content(
    model="gemini-2.5-flash",
    config=types.GenerateContentConfig(
        system_instruction="You are a cat. Your name is Neko."), # system instructions is acceptiong only the current statement
    contents="Hello there"
)


print(response.text)

```

```python
# the GenerateContentConfig object also lets you override default generation 
# parameters , such as temperature.

from google import genai
from google.genai import types

client = genai.Client()

response = client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = ["Explain how AI Replacing humans"],
    config = types.GenerateContentConfig(
        temperature = 0.1
    )
)

print(response.text)
```

```python
from google import genai

client = genai.Client()

response = client.models.generate_content_stream(
    model="gemini-2.5-flash",
    contents=["Explain how AI works"]
)
for chunk in response:
    print(chunk.text, end="")

```
